## Merge the four parquet files into one pandas dataframe

```python
import pandas as pd

grid = pd.read_parquet("iowa_grid_lookup.parquet")
panel_1 = pd.read_parquet("00_05_nitrogen_surplus_iowa_grid_panel.parquet")
panel_2 = pd.read_parquet("06_11_nitrogen_surplus_iowa_grid_panel.parquet")
panel_3 = pd.read_parquet("12_17_nitrogen_surplus_iowa_grid_panel.parquet")
panel = pd.concat([panel_1, panel_2, panel_3], ignore_index=True)
```

```python
print("grid rows:", len(grid), "| pixel_id dtype:", grid["pixel_id"].dtype)
print("panel rows:", len(panel), "| pixel_id dtype:", panel["pixel_id"].dtype)
print("years available:", sorted(panel["year"].unique()))
print("lat range:", grid["lat"].min(), "to", grid["lat"].max())
print("lon range:", grid["lon"].min(), "to", grid["lon"].max())
```

```python
merged = panel.merge(grid, on="pixel_id", how="inner")
print("rows after merge:", len(merged))
print(merged.sample(10))
```
